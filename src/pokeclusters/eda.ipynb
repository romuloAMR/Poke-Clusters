```python
stats_cols = ['HP', 'Attack', 'Defense', 'SP_Attack', 'SP_Defense', 'Speed']

type_stats = df.groupby('Type1')[stats_cols].mean()

type_stats['Total_Avg'] = type_stats.sum(axis=1)

type_stats = type_stats.sort_values('Total_Avg', ascending=False)

plt.figure(figsize=(14, 10))
sns.heatmap(type_stats[stats_cols], 
            annot=True, 
            cmap='YlOrRd', 
            fmt='.0f',
            linewidths=0.5,
            cbar_kws={'label': 'Mean value'})
plt.title('Mean statistics by pokemon type', fontsize=16, pad=20)
plt.xlabel('Attribute')
plt.ylabel('Type')
plt.tight_layout()
plt.show()
```

## Feature Engineering and Selection

```python
df["Type1"].unique()
```

```python
tcg_mapping = {
    'Grass': 'Grass',
    'Bug': 'Grass',
    'Fire': 'Fire',
    'Water': 'Water',
    'Ice': 'Water',
    'Electric': 'Lightning',
    
    'Psychic': 'Psychic',
    'Ghost': 'Psychic',
    'Fairy': 'Psychic', 
    'Dark': 'Darkness',
    'Poison': 'Darkness',
    
    'Fighting': 'Fighting',
    'Rock': 'Fighting',
    'Ground': 'Fighting',    
    'Steel': 'Metal',
    
    'Dragon': 'Dragon',
    'Normal': 'Colorless',
    'Flying': 'Colorless'
}

df["New_Type1"] = df["Type1"].map(tcg_mapping)
df['New_Type2'] = df['Type2'].map(tcg_mapping)
```

```python
dummies1 = pd.get_dummies(df['New_Type1'], dtype=int)
dummies2 = pd.get_dummies(df['New_Type2'], dtype=int)

df_encoded = dummies1.add(dummies2, fill_value=0)
df_encoded = (df_encoded > 0).astype(int)

df = pd.concat([df, df_encoded], axis=1)

df[['Name'] + list(df_encoded.columns)].head()
```

```python
battle_efforts = [col for col in df.columns if col.startswith('E_')]
df = df.drop(columns=battle_efforts)

df = df.drop(columns=['No','Generation', 'Height', 'Ability_Hidden', 'Color', 'Gender_Male', 'Gender_Female', 'Gender_Unknown', 'Egg_Steps', 'Experience_Type', 'Base_Experience', 'Egg_Group1', 'Egg_Group2', 'Type1', 'Type2', 'Num_Abilities', 'Weight', 'Branch_Code', 'Total', 'Ability1', 'Ability2', 'Region_Form', 'New_Type1', 'New_Type2'])

df.info()
```

```python
df.to_csv("../data/pokedex_post_eda.csv", index=False)
```
